```python
# Initializing the general function to perform date operations

def split_date(df):
    df['Date'] = pd.to_datetime(df['Date']) # Since the date feature is a string data type, use this to convert it to datetime datatype
    df['Year'] = df.Date.dt.year # Create the Year feature from the date feature
    df['Month'] = df.Date.dt.month # Create the Month feature from the date feature
    df['Day'] = df.Date.dt.day # Create the Day feature from the date feature
    df['WeekOfYear'] = df.Date.dt.isocalendar().week
    
split_date(rossman_sales_train)
split_date(rossman_sales_test)

# You can check using these: 
# rossman_sales_train.head(6)
# rossman_sales_test.head(6)
```

FEATURE ENGINEERING ON COMPETITION

1. CompetitionDistance - distance in meters to the nearest competitor store

2. CompetitionOpenSince[Month/Year] - gives the approximate year and month of the time the nearest competitor was opened

```python
# Use the CompetitionOpenSince[Month/Year] columns from the store_df to compute the number of months for which a competitor has been open near the store

def comp_months(df):
    df['CompetitionOpen'] = 12 * (df.Year - df.CompetitionOpenSinceYear) + (df.Month - df.CompetitionOpenSinceMonth)
    df['CompetitionOpen'] = df['CompetitionOpen'].map(lambda x: 0 if x < 0 else x).fillna(0)
    
    return df
    
```

The CompetitionOpen column represents the total number of months a competitor has been active and operating near a specific store at the time of a given sales record.

Time-Aware Metric: It is not a fixed number. It changes dynamically depending on the date of the sales record.

Zero Value Meaning: A value of 0 means the competitor has not opened yet, opened in the current month, or the opening date is unknown.

Data Unit: The final output is an integer representing time measured strictly in months.

This is done because:
   1. Raw dates (like 2025 or March) are hard for algorithms to interpret. A single continuous number (like 14 months) is much easier for a machine learning model to read.

   2. Retail stores often see a drop in sales when a competitor first opens nearby. This column helps the model learn if a competitor is brand new (low number) or long-established (high number).


THE LAMBDA FUNCTION: 

1. Handle Future Dates: 

If the competitor opens after the transaction date, the subtraction results in a negative number. The lambda function forces any negative values to 0 (meaning 0 months open).Handle Missing Data: The .fillna(0) converts any NaN (missing) values to 0, which happens if the opening month or year data was originally blank.

2. Handle Missing Data: 

The .fillna(0) converts any NaN (missing) values to 0, which happens if the opening month or year data was originally blank.

FEATURE ENGINEERING ON PROMOTION

Adding some additional columns to indicate how long a store has been running Promo2 and whether a new round of Promo2 starts in the current month

Promo - indicates whether a store is running a promo on that day

Promo2 - Promo2 is a continuing and consecutive promotion for some stores: 0 = store is not participating, 1 = store is participating

Promo2Since[Year/Week] - describes the year and calendar week when the store started participating in Promo2

PromoInterval - describes the consecutive intervals Promo2 is started, naming the months the promotion is started anew. E.g. "Feb,May,Aug,Nov" means each round starts in February, May, August, November of any given year for that store

Rossmann store runs a repeating promotion called Promo2. The things we want to know about each row (a store on a specific date) are:

How long has Promo2 been running at this store? (in months)

Is this month one where a new round of Promo2 just kicked off?

```python
def check_promo_month(row):
    month2str = {
        1:'Jan',
        2:'Feb',
        3:'Mar',
        4:'Apr',
        5:'May',
        6:'Jun',
        7:'Jul',
        8:'Aug',
        9:'Sept',
        10:'Oct',
        11:'Nov',
        12:'Dec'
    }
    
    try:
        # Dataset stores a string like "Jan,Apr,Jul,Oct" telling you which months a new promo round begins. This line breaks that string into a list: ['Jan', 'Apr', 'Jul', 'Oct'].
        
        months = (row['PromoInterval'] or '').split(',')
        
        # If Promo2 is already running here, AND this row's month is one of the months listed in that string — say yes (1). Otherwise, say no (0)
        
        if row['Promo2Open'] and month2str[row['Month']] in months:
            return 1
        else:
            return 0
        
    except Exception:
        return 0
    
def promo_cols(df):
    # Months since Promo2 was open
    # take the year gap, turn it into months (×12), then add the leftover week gap. It's a rough "how many months has this promo been live" number.
    # If the number comes out negative (meaning Promo2 hadn't started yet on this date), just call it 0.
    # If it's missing, call it 0 too.
    # Then multiply by Promo2 (which is 1 if the store participates, 0 if not) — so stores that never signed up for Promo2 automatically get 0, no matter what.
    
    df['Promo2Open'] = 12 * (df.Year - df.Promo2SinceYear) + (df.WeekOfYear - df.Promo2SinceWeek)* 7 / 30.5
    df['Promo2Open'] = df['Promo2Open'].map(lambda x: 0  if x < 0 else x).fillna(0) * df['Promo2']
    
    # Checking whether a new round of promotions was started in the current month
    df['IsPromo2Month'] = df.apply(check_promo_month, axis=1) * df['Promo2']
    
    return df
```

```python
# FOR QUICK INSPECTION

rossman_sales_train= promo_cols(rossman_sales_train)
rossman_sales_test = promo_cols(rossman_sales_test)

# rossman_sales_train
```

```python
# SEPARATING THE INPUT FEATURES FROM THE TARGET FEATURES

input_cols = [
    'Store', 'DayOfWeek', 'Promo', 'Promo2',
    'StateHoliday', 'SchoolHoliday', 'StoreType', 'Assortment',
    'CompetitionDistance', 'Year', 'Month', 'Day','WeekOfYear', 'CompetitionOpen', 'Promo2Open', 'IsPromo2Month'
    ]

target_col = 'Sales'

inputs = rossman_sales_train[input_cols].copy()
targets = rossman_sales_train[target_col].copy()

# Create a copy for the test dataset

test_inputs = rossman_sales_test[input_cols].copy()
```

CHECKING THE NUMERIC AND CATEGORICAL FEATURES

```python
numeric_cols = [
    'Store', 'Promo', 'Promo2',
    'SchoolHoliday','CompetitionDistance', 'Year', 
    'Month', 'Day','WeekOfYear', 
    'CompetitionOpen', 'Promo2Open', 'IsPromo2Month'
    ]

categorical_cols = ['StateHoliday', 'StoreType', 'Assortment', 'DayOfWeek']
```

IMPUTING MISSING NUMERICAL DATA

```python
inputs[numeric_cols].isna().sum()
```

```python
test_inputs[numeric_cols].isna().sum()
```

```python
# You can fill the competition distance with the maximum value to show that it is really far away
max_distance = inputs['CompetitionDistance'].max() # Find the maximum  value in training
print('THE MAXIMUM DISTANCE IS: ', max_distance)

inputs['CompetitionDistance']=inputs['CompetitionDistance'].fillna(max_distance)
test_inputs['CompetitionDistance']=test_inputs['CompetitionDistance'].fillna(max_distance)
```

SCALING THE NUMERIC COLUMNS ON THE 0-1 RANGE

```python
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder

# Initializing and fitting the scaler only on the training inputs
scaler = MinMaxScaler().fit(inputs[numeric_cols])

# Now tranforming both on the training and test inputs
inputs[numeric_cols] = scaler.transform(inputs[numeric_cols])
test_inputs[numeric_cols] = scaler.transform(test_inputs[numeric_cols])
```

ENCODING CATEGORICAL COLUMNS

```python
# TO CHECK THE COLUMNS THAT HAVE MIXED DATA TYPES

#for col in categorical_cols:
    #types = inputs[col].apply(type).unique()
    #if len(types) > 1:
        #print(f"Column '{col}' has mixed types: {types}")

# Cast the  categorical columns explicitly to string
inputs[categorical_cols] = inputs[categorical_cols].astype(str)

# Force test data categorical columns to strings to match training data format
test_inputs[categorical_cols] = test_inputs[categorical_cols].astype(str)


# Initializing the encoder and fitting the training categorical columns
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore').fit(inputs[categorical_cols])

# GETTING THE NEW ENCODED FEATURE NAMES
encoded_cols = list(encoder.get_feature_names_out(categorical_cols))
encoded_cols

# Now transforming the encoded columns both on the training and test inputs
inputs[encoded_cols] = encoder.transform(inputs[categorical_cols])
test_inputs[encoded_cols] = encoder.transform(test_inputs[categorical_cols])
```

FINAL PREPARED TRAINING AND TESTING INPUT SETS

```python
X_train = inputs[numeric_cols + encoded_cols]
X_test = test_inputs[numeric_cols + encoded_cols]
```
